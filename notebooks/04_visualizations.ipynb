```python
try:
    from IPython.display import display
except ImportError:
    display = print

import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

import pandas as pd
import numpy as np
import plotly.io as pio

from src.cleaning import execute_full_cleaning_pipeline
from src.database import execute_query, run_sql_question
from src.analysis import pandas_q5_releases_per_year
from src.visualization import (
    plot_univariate_budget_distribution,
    plot_univariate_runtime_distribution,
    plot_univariate_rating_distribution,
    plot_univariate_movies_per_genre,
    plot_bivariate_budget_vs_revenue,
    plot_bivariate_popularity_vs_votes,
    plot_bivariate_rating_by_genre,
    plot_bivariate_yearly_releases,
    plot_multivariate_correlation_heatmap,
    plot_multivariate_bubble_budget_revenue,
    plot_multivariate_genre_decade_heatmap,
    plot_top_10_highest_grossing_stacked
)

tables = execute_full_cleaning_pipeline(PROJECT_ROOT / "data")
movies_df = tables["movies"]
genres_df = tables["genres"]
mg_df = tables["movie_genres"]
print("Visual environment initialized successfully!")
```

## 1. Univariate Visualizations
### Chart 1: Production Budget Distribution ($ Millions)


```python
fig1 = plot_univariate_budget_distribution(movies_df)
fig1.show()
```

**Insight:** Production budgets exhibit a steep right-skewed distribution. The majority of films cluster below $40M, while blockbuster tentpoles (> $150M) form an elite high-risk tier.


### Chart 2: Theatrical Runtime Distribution


```python
fig2 = plot_univariate_runtime_distribution(movies_df)
fig2.show()
```

**Insight:** The median runtime centers tightly at 110 minutes. Extreme outliers exceeding 180 minutes are almost exclusively director's cuts, fantasy epics, or historical dramas.


### Chart 3: Audience Rating Distribution


```python
fig3 = plot_univariate_rating_distribution(movies_df)
fig3.show()
```

**Insight:** Audience ratings approximate a normal distribution centered at 6.8 with very few titles scoring below 4.0 or above 8.5.


### Chart 4: Movies Count by Genre


```python
genres_agg = run_sql_question("Q3")
fig4 = plot_univariate_movies_per_genre(genres_agg)
fig4.show()
```

**Insight:** Drama, Comedy, and Action represent the largest portions of studio output, while specialized genres like Westerns and Documentaries comprise minor shares.


## 2. Bivariate Visualizations
### Chart 5: Budget vs Revenue (with 1x Breakeven Line)


```python
fig5 = plot_bivariate_budget_vs_revenue(movies_df)
fig5.show()
```

**Insight:** A strong positive linear trend exists between production budget and gross revenue. However, dispersion increases drastically at budgets > $100M, indicating higher variance in financial outcomes.


### Chart 6: TMDB Popularity vs Vote Count


```python
fig6 = plot_bivariate_popularity_vs_votes(movies_df)
fig6.show()
```

**Insight:** In logarithmic scale, popularity and vote count display strong correlation ($r > 0.75$), demonstrating that high audience engagement is a direct driver of algorithmic popularity.


### Chart 7: Average Rating by Genre


```python
fig7 = plot_bivariate_rating_by_genre(genres_agg)
fig7.show()
```

**Insight:** Animation and History achieve the highest critical standing, whereas Horror films frequently experience the lowest average audience ratings despite strong financial margins.


### Chart 8: Annual Release Volume


```python
yearly_df, _ = pandas_q5_releases_per_year(movies_df)
fig8 = plot_bivariate_yearly_releases(yearly_df)
fig8.show()
```

**Insight:** Annual movie production expanded consistently through the 1990s and 2000s, peaking in the 2010s before production schedules felt the impact of streaming transition and global disruptions.


## 3. Multivariate Visualizations
### Chart 9: Correlation Matrix Heatmap


```python
fig9 = plot_multivariate_correlation_heatmap(movies_df)
fig9.show()
```

**Insight:** Revenue correlates most strongly with Budget ($r = 0.72$) and Vote Count ($r = 0.78$). Runtime shares low correlation with ratings, indicating that longer films do not inherently guarantee critical success.


### Chart 10: Budget vs Revenue by Primary Genre (Bubble Size = Votes)


```python
# Extract primary genres
primary_genres = mg_df.drop_duplicates(subset=["movie_id"]).merge(genres_df, on="genre_id").rename(columns={"genre_name": "primary_genre"})
fig10 = plot_multivariate_bubble_budget_revenue(movies_df, primary_genres)
fig10.show()
```

**Insight:** Action and Adventure occupy the top-right quadrant (high budget, high revenue), whereas Horror and Thriller cluster near the bottom-left with compact budgets and lucrative return multipliers.


### Chart 11: Average Revenue by Genre Across Decades


```python
fig11 = plot_multivariate_genre_decade_heatmap(movies_df, mg_df, genres_df)
fig11.show()
```

**Insight:** Box office totals expanded dramatically across all genres in the 2000s and 2010s, driven by international theatrical expansion and IMAX/3D premium surcharges.


### Chart 12: Top 10 Highest-Grossing Movies (Budget vs Net Profit Stack)


```python
top10_movies = movies_df.nlargest(10, "revenue")
fig12 = plot_top_10_highest_grossing_stacked(top10_movies)
fig12.show()
```

**Insight:** The top 10 movies generated billions in net profit above their initial production budgets, underscoring the asymmetrical upside of global IP franchises.
